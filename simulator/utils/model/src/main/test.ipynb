```python
import logging
from pathlib import Path

import enoslib as en

en.init_logging(level=logging.INFO)
en.check()



conf = (
    en.G5kConf.from_settings(job_name="Tuto", job_type=[])
    .add_machine(roles=["master"], cluster="econome", nodes=1)
    
    .add_machine(roles=["slave"], cluster="econome", nodes=1)
    .add_machine(roles=["slave"], cluster="econome", nodes=1)
)
provider = en.G5k(conf)
roles, networks = provider.init()
roles = en.sync_info(roles, networks)
```

```python


netem = en.NetemHTB()
(
    netem.add_constraints(
        src=roles["paris"],
        dest=roles["londres"],
        delay="500ms",
        rate="1gbit",
        symmetric=True,
    )
    .add_constraints(
        src=roles["paris"],
        dest=roles["berlin"],
        delay="500ms",
        rate="1gbit",
        symmetric=True,
    )
    .add_constraints(
        src=roles["londres"],
        dest=roles["berlin"],
        delay="500ms",
        rate="1gbit",
        symmetric=True,
    )
)
netem.deploy()
netem.validate()
```
